```python
import os
from pathlib import Path

repo_root = Path.cwd()
if not (repo_root / 'services').exists():
    raise RuntimeError('Run this notebook from the repository root so the services package is importable.')

required_env = [
    'GOOGLE_API_KEY',
    'SECRET_KEY',
    'API_KEY',
    'EBAY_CLIENT_ID',
    'EBAY_CLIENT_SECRET',
    'EBAY_RU_NAME',
]

missing = [key for key in required_env if not os.getenv(key)]
if missing:
    print('Missing environment variables:')
    for key in missing:
        print(f' - {key}')
    print('\nSet them before instantiating external services.')
else:
    print('All required environment variables are present.')
```

```python
import os

from services.vision_service import VisionService
from services.conversation_orchestrator import ConversationOrchestrator

google_api_key = os.getenv('GOOGLE_API_KEY')
secret_key = os.getenv('SECRET_KEY')
api_key = os.getenv('API_KEY')

if not google_api_key:
    raise RuntimeError('GOOGLE_API_KEY is not set.')

vision_service = VisionService(api_key=google_api_key)
conversation_orchestrator = ConversationOrchestrator()

print('VisionService initialized successfully.')
print('ConversationOrchestrator initialized successfully.')
print(f'GOOGLE_API_KEY configured: {bool(google_api_key)}')
print(f'SECRET_KEY configured: {bool(secret_key)}')
print(f'API_KEY configured: {bool(api_key)}')
```

```python
sample_item = {
    'item_name': 'Wireless Headphones',
    'category': 'Electronics',
    'condition': 'Used',
}

session = conversation_orchestrator.start_conversation(item_id='demo-item-1', initial_data=sample_item)
print('Conversation session created.')
print(f'Session ID: {session.session_id}')
print(f'Current question: {session.current_question}')

updated_state = conversation_orchestrator.process_answer(
    session.session_id,
    'Yes, they work normally and include the original case.'
)

print('Answer processed.')
print(f'Confidence: {updated_state.confidence:.2f}')
print(f'Known fields: {sorted(updated_state.known_fields.keys())}')
```
